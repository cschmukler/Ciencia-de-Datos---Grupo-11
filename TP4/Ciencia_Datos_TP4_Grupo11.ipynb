##Creación de Pobre y No Pobre

```python
import numpy as np

# ======================
# Cálculo de pobreza con base unificada EPH_limpio
# ======================

# 1. Función de adultos equivalentes robusta a NAs
def adulto_equiv(edad, sexo):
    if pd.isna(edad) or pd.isna(sexo):
        return np.nan
    edad = int(edad)
    sexo = int(sexo)

    if edad < 1: return 0.35
    elif edad == 1: return 0.37
    elif edad == 2: return 0.46
    elif edad == 3: return 0.51
    elif edad == 4: return 0.55
    elif edad == 5: return 0.60
    elif edad == 6: return 0.64
    elif edad == 7: return 0.66
    elif edad == 8: return 0.68
    elif edad == 9: return 0.69
    elif edad == 10: return 0.70 if sexo == 2 else 0.79
    elif edad == 11: return 0.72 if sexo == 2 else 0.82
    elif edad == 12: return 0.74 if sexo == 2 else 0.85
    elif edad == 13: return 0.76 if sexo == 2 else 0.90
    elif edad == 14: return 0.76 if sexo == 2 else 0.96
    elif edad == 15: return 0.77 if sexo == 2 else 1.00
    elif edad == 16: return 0.77 if sexo == 2 else 1.03
    elif edad == 17: return 0.77 if sexo == 2 else 1.04
    elif 18 <= edad <= 29: return 0.76 if sexo == 2 else 1.02
    elif 30 <= edad <= 45: return 0.77 if sexo == 2 else 1.00
    elif 46 <= edad <= 60: return 0.76 if sexo == 2 else 1.00
    elif 61 <= edad <= 75: return 0.67 if sexo == 2 else 0.83
    else: return 0.63 if sexo == 2 else 0.74

# 2. Aplicamos la función
EPH_limpio["adulto_equiv"] = EPH_limpio.apply(
    lambda row: adulto_equiv(row["CH06"], row["CH04"]), axis=1
)

# 3. Adultos equivalentes por hogar
EPH_limpio["ad_equiv_hogar"] = (
    EPH_limpio.groupby(["CODUSU","NRO_HOGAR"])["adulto_equiv"].transform("sum")
)

# 4. Canasta Básica Total (CBT) por año
CBT = {2005: 205.07, 2025: 365177}

# 5. Ingreso necesario
EPH_limpio["ingreso_necesario"] = (
    EPH_limpio["AÑO"].map(CBT) * EPH_limpio["ad_equiv_hogar"]
)

# 6. Variable pobre
EPH_limpio["pobre"] = (
    (EPH_limpio["ITF"] < EPH_limpio["ingreso_necesario"])
).astype(int)

# 7. Resultados
resultados = (
    EPH_limpio.groupby("AÑO")["pobre"]
    .agg(["sum","count"])
    .rename(columns={"sum":"pobres","count":"total"})
)
resultados["%"] = 100 * resultados["pobres"] / resultados["total"]

print(resultados)

EPH_limpio
```

##Recapitulando las cosas necesarias del TP3

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Diccionarios para guardar los datos
train_dfs = {}
test_dfs = {}
X_trains = {}
y_trains = {}
X_tests = {}
y_tests = {}

for año in [2005, 2025]:
    # Filtramos la base por año (columna en mayúscula)
    df_año = respondieron[respondieron['AÑO'] == año].copy()

    # Definimos variable dependiente y variables independientes
    y = df_año['pobre']
    X = df_año.drop(columns=['pobre', 'ITF', 'AÑO'])  # excluimos ingreso y año

    # Agregamos columna de 1 para intercepto
    X.insert(0, 'intercept', 1)

    # Dividimos en entrenamiento (70%) y prueba (30%)
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, train_size=0.7, random_state=444
    )

    # Guardamos en diccionarios
    train_dfs[año] = pd.concat([X_train, y_train], axis=1)
    test_dfs[año] = pd.concat([X_test, y_test], axis=1)
    X_trains[año] = X_train
    y_trains[año] = y_train
    X_tests[año] = X_test
    y_tests[año] = y_test

# Ver ejemplo de la base de entrenamiento de 2005
print("Train 2005 shape:", train_dfs[2005].shape)
print("Test 2005 shape:", test_dfs[2005].shape)

X_trains
```

```python
from scipy.stats import ttest_ind
import numpy as np
import pandas as pd

vars_x = [
    "V18_M", "CH04", "CH06", "CH07", "CH08",
    "NIVEL_ED", "ESTADO", "CAT_INAC", "IPCF",
    "PP08D1", "CH09", "PP04C", "CH12", "PP05C_2", "PP07C"
]

diff_means = {}

for año in [2005, 2025]:
    train = train_dfs[año]
    test = test_dfs[año]

    # Variables que están en train y test y son numéricas
    vars_validas = [v for v in vars_x if v in train.columns and v in test.columns
                    and pd.api.types.is_numeric_dtype(train[v])]

    # Medias
    mean_train = train[vars_validas].mean()
    mean_test = test[vars_validas].mean()

    # Diferencia
    diff = mean_train - mean_test

    # Test t
    p_values = []
    for var in vars_validas:
        try:
            stat, p = ttest_ind(train[var].dropna(), test[var].dropna(), equal_var=False)
            p_values.append(p)
        except:
            p_values.append(np.nan)

    # Armamos tabla alineada
    tabla = pd.DataFrame({
        "variable": vars_validas,
        "media_train": mean_train.values,
        "media_test": mean_test.values,
        "diferencia": diff.values,
        "p_value": p_values
    }).set_index("variable").round(3)

    diff_means[año] = tabla

    print(f"\nDiferencia de medias - Año {año}")
    display(tabla)

    # Mostrar qué variables fueron excluidas
    excluidas = [v for v in vars_x if v not in vars_validas]
    print(f"Año {año} - variables excluidas (no numéricas o no presentes): {excluidas}")
```

```python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LogisticRegression

# Grid de lambdas: λ = 10^n, n ∈ {-5, ..., 5}
n_values = np.arange(-5, 6)
lambdas = 10.0 ** n_values
C_values = 1 / lambdas   # C es la inversa de la penalización

# 5-fold CV
kf = KFold(n_splits=5, shuffle=True, random_state=42)

errors_ridge = []   # cada elemento: array de 5 errores (uno por fold)
errors_lasso = []

for C in C_values:
    # ---------- Ridge (L2) ----------
    ridge_cv = LogisticRegression(
        penalty="l2",
        C=C,
        solver="liblinear",
        max_iter=200
    )
    acc_ridge = cross_val_score(
        ridge_cv, X_train_std, y_train,
        cv=kf, scoring="accuracy"
    )
    errors_ridge.append(1 - acc_ridge)   # error = 1 - accuracy

    # ---------- LASSO (L1) ----------
    lasso_cv = LogisticRegression(
        penalty="l1",
        C=C,
        solver="liblinear",
        max_iter=200
    )
    acc_lasso = cross_val_score(
        lasso_cv, X_train_std, y_train,
        cv=kf, scoring="accuracy"
    )
    errors_lasso.append(1 - acc_lasso)

# Pasamos a arrays: forma (n_lambdas, 5)
errors_ridge = np.array(errors_ridge)
errors_lasso = np.array(errors_lasso)

# Error promedio por λ
mean_err_ridge = errors_ridge.mean(axis=1)
mean_err_lasso = errors_lasso.mean(axis=1)

# Índice del mejor λ (mínimo error promedio)
idx_best_ridge = np.argmin(mean_err_ridge)
idx_best_lasso = np.argmin(mean_err_lasso)

lambda_cv_ridge = lambdas[idx_best_ridge]
lambda_cv_lasso = lambdas[idx_best_lasso]

print("λ^cv Ridge:", lambda_cv_ridge, "   (C =", C_values[idx_best_ridge], ")")
print("λ^cv LASSO:", lambda_cv_lasso, "   (C =", C_values[idx_best_lasso], ")")
```

```python
import matplotlib.pyplot as plt

# Para que el eje x quede prolijo en log10(λ)
x_pos = np.log10(lambdas)
x_labels = [f"1e{n}" for n in n_values]

plt.figure(figsize=(12,5))

# --------- Ridge ----------
plt.subplot(1,2,1)
plt.boxplot(errors_ridge.T, positions=x_pos)
plt.axvline(np.log10(lambda_cv_ridge), color="red", linestyle="--",
            label=r"$\lambda^{cv}$")
plt.xticks(x_pos, x_labels, rotation=45)
plt.xlabel("λ")
plt.ylabel("Error de clasificación (validación)")
plt.title("Ridge (L2) – 5-fold CV")
plt.grid(alpha=0.3)
plt.legend()

# --------- LASSO ----------
plt.subplot(1,2,2)
plt.boxplot(errors_lasso.T, positions=x_pos)
plt.axvline(np.log10(lambda_cv_lasso), color="red", linestyle="--",
            label=r"$\lambda^{cv}$")
plt.xticks(x_pos, x_labels, rotation=45)
plt.xlabel("λ")
plt.ylabel("Error de clasificación (validación)")
plt.title("LASSO (L1) – 5-fold CV")
plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

plt.figure()

for nombre, modelo in modelos.items():
    y_prob = modelo.predict_proba(X_test)[:,1]
    fpr, tpr, _ = roc_curve(y_test, y_prob)
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, label=f"{nombre} (AUC = {roc_auc:.3f})")

plt.plot([0,1],[0,1],'k--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Curvas ROC - Comparación de Modelos")
plt.legend()
plt.show()
```
