```python
import numpy as np

# ======================
# Cálculo de pobreza con base unificada EPH_limpio
# ======================

# 1. Función de adultos equivalentes robusta a NAs
def adulto_equiv(edad, sexo):
    if pd.isna(edad) or pd.isna(sexo):
        return np.nan
    edad = int(edad)
    sexo = int(sexo)

    if edad < 1: return 0.35
    elif edad == 1: return 0.37
    elif edad == 2: return 0.46
    elif edad == 3: return 0.51
    elif edad == 4: return 0.55
    elif edad == 5: return 0.60
    elif edad == 6: return 0.64
    elif edad == 7: return 0.66
    elif edad == 8: return 0.68
    elif edad == 9: return 0.69
    elif edad == 10: return 0.70 if sexo == 2 else 0.79
    elif edad == 11: return 0.72 if sexo == 2 else 0.82
    elif edad == 12: return 0.74 if sexo == 2 else 0.85
    elif edad == 13: return 0.76 if sexo == 2 else 0.90
    elif edad == 14: return 0.76 if sexo == 2 else 0.96
    elif edad == 15: return 0.77 if sexo == 2 else 1.00
    elif edad == 16: return 0.77 if sexo == 2 else 1.03
    elif edad == 17: return 0.77 if sexo == 2 else 1.04
    elif 18 <= edad <= 29: return 0.76 if sexo == 2 else 1.02
    elif 30 <= edad <= 45: return 0.77 if sexo == 2 else 1.00
    elif 46 <= edad <= 60: return 0.76 if sexo == 2 else 1.00
    elif 61 <= edad <= 75: return 0.67 if sexo == 2 else 0.83
    else: return 0.63 if sexo == 2 else 0.74

# 2. Aplicamos la función
EPH_limpio["adulto_equiv"] = EPH_limpio.apply(
    lambda row: adulto_equiv(row["CH06"], row["CH04"]), axis=1
)

# 3. Adultos equivalentes por hogar
EPH_limpio["ad_equiv_hogar"] = (
    EPH_limpio.groupby(["CODUSU","NRO_HOGAR"])["adulto_equiv"].transform("sum")
)

# 4. Canasta Básica Total (CBT) por año
CBT = {2005: 205.07, 2025: 365177}

# 5. Ingreso necesario
EPH_limpio["ingreso_necesario"] = (
    EPH_limpio["AÑO"].map(CBT) * EPH_limpio["ad_equiv_hogar"]
)

# 6. Variable pobre
EPH_limpio["pobre"] = (
    (EPH_limpio["ITF"] < EPH_limpio["ingreso_necesario"])
).astype(int)

# 7. Resultados
resultados = (
    EPH_limpio.groupby("AÑO")["pobre"]
    .agg(["sum","count"])
    .rename(columns={"sum":"pobres","count":"total"})
)
resultados["%"] = 100 * resultados["pobres"] / resultados["total"]

print(resultados)

```

```python

import pandas as pd
import numpy as np

# --------- Funciones de limpieza y cálculo ---------
def _clean_tripleta(ch12, ch13, ch14):
    # Forzar a numérico
    ch12 = pd.to_numeric(ch12, errors="coerce")
    ch13 = pd.to_numeric(ch13, errors="coerce")
    ch14 = pd.to_numeric(ch14, errors="coerce")

    # CH12: nivel educativo (filtrar códigos raros)
    # válidos típicos: 1..9 (ajustable). Marcamos 0,10,99 como NaN
    if ch12 in [0, 10, 99]:
        ch12 = np.nan

    # CH13: finalizó (1 sí, 2 no). 0 no válido
    if ch13 == 0:
        ch13 = np.nan

    # CH14: último año aprobado (0..12/13), 98/99 = especiales
    if ch14 in [98, 99]:
        ch14 = np.nan
    # Recorte conservador de outliers (opcional): 0..20
    if pd.notna(ch14) and (ch14 < 0 or ch14 > 20):
        ch14 = np.nan

    # Convertir a enteros cuando se pueda
    if pd.notna(ch12): ch12 = int(ch12)
    if pd.notna(ch13): ch13 = int(ch13)
    if pd.notna(ch14): ch14 = int(ch14)

    return ch12, ch13, ch14

def calcular_educ_row(row):
    ch12, ch13, ch14 = _clean_tripleta(row["CH12"], row["CH13"], row["CH14"])

    # Si falta nivel, no podemos calcular
    if pd.isna(ch12):
        return np.nan

    # Si falta CH13 (finalizó) y tengo CH14, asumo INCOMPLETO
    # Si falta CH13 y también CH14, no podemos calcular
    if pd.isna(ch13):
        if pd.notna(ch14):
            ch13 = 2  # tratar como "no finalizado"
        else:
            return np.nan

    # Si falta CH14 y ch13 indica "no finalizado", no sabemos cuántos años -> NaN
    if ch13 == 2 and pd.isna(ch14):
        return np.nan

    # Lógica de años (ajustable a tu diseño):
    # Primario completo = 6; Secundario/EGB/Polimodal completo = 12;
    # Terciario completo ~ 15 (12 + 3); Universitario completo ~ 17 (12 + 5);
    # Posgrado: sumar encima de universitario
    educ = np.nan

    if ch12 == 1:  # Jardín
        educ = 0

    elif ch12 == 2:  # Primario
        educ = 6 if ch13 == 1 else ch14

    elif ch12 in [3, 4, 5]:  # EGB / Secundario / Polimodal
        educ = 12 if ch13 == 1 else (6 + ch14)

    elif ch12 == 6:  # Terciario
        educ = 15 if ch13 == 1 else (12 + ch14)

    elif ch12 == 7:  # Universitario
        educ = 17 if ch13 == 1 else (12 + ch14)

    elif ch12 == 8:  # Posgrado
        # Si CH14 está para posgrado, lo sumamos encima de universitario
        if pd.notna(ch14):
            educ = 17 + ch14
        else:
            # Si no sabemos cuántos años de posgrado, al menos universitario completo
            educ = 17 if ch13 == 1 else np.nan

    elif ch12 == 9:  # Educación especial
        educ = np.nan

    return educ

# Calcula educ con la limpieza embebida
EPH_limpio["educ"] = EPH_limpio.apply(calcular_educ_row, axis=1)

#Dejar educ como entero "nullable" cuando sea entero
if EPH_limpio["educ"].dropna().mod(1).eq(0).all():
    EPH_limpio["educ"] = EPH_limpio["educ"].astype("Int64")

# --------- Estadística descriptiva
stats = EPH_limpio["educ"].describe()[["mean","std","min","50%","max"]]
print(stats)

# casos con educ muy alto para sanity-check
outliers = EPH_limpio.loc[EPH_limpio["educ"].fillna(-1) > 25, ["CH12","CH13","CH14","educ"]]
print("\nCasos con educ > 25 (para revisar):", len(outliers))
print(outliers.head(10))
```

```python
# === PCA con las 6 variables (edad, edad2, educ, ITF, MIEMBROS_HOGAR, horastrab) ===
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1) Preparación de variables
# --------------------------------------------------
df = EPH_limpio.copy()

# Miembros del hogar (si no existe, lo calculamos desde la base individual)
if "MIEMBROS_HOGAR" not in df.columns:
    if all(c in df.columns for c in ["CODUSU","NRO_HOGAR","COMPONENTE"]):
        df["MIEMBROS_HOGAR"] = (
            df.groupby(["CODUSU","NRO_HOGAR"])["COMPONENTE"].transform("nunique")
        ).astype("Int64")
    else:
        raise ValueError("Faltan columnas para calcular MIEMBROS_HOGAR (CODUSU/NRO_HOGAR/COMPONENTE).")

# Horas trabajadas (si no existe, la calculamos y recortamos a 84)
if "horastrab" not in df.columns:
    h1 = df["PP3E_TOT"].where(df["PP3E_TOT"].between(0, 100), np.nan)
    h2 = df["PP3F_TOT"].where(df["PP3F_TOT"].between(0, 100), np.nan)
    df["horastrab"] = (h1.fillna(0) + h2.fillna(0)).clip(upper=84)

# edad y edad2
df["edad"]  = pd.to_numeric(df["CH06"], errors="coerce")
df["edad2"] = df["edad"]**2

# educ (usamos NIVEL_ED ya recodificada)
df["educ"] = pd.to_numeric(df["NIVEL_ED"], errors="coerce")

# --------------------------------------------------
# 2) Dataset para PCA (dropna en las 6 variables)
# --------------------------------------------------
vars_pca = ["edad", "edad2", "educ", "ITF", "MIEMBROS_HOGAR", "horastrab"]
X = df[vars_pca].astype(float)
mask = X.notna().all(axis=1)
X = X[mask]
df_plot = df.loc[mask].copy()   # para recuperar AÑO en el gráfico

# --------------------------------------------------
# 3) Estandarización y PCA
# --------------------------------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2, random_state=0)
X_pca = pca.fit_transform(X_scaled)

scores = pd.DataFrame(X_pca, columns=["PC1","PC2"], index=X.index)

# Varianza explicada y loadings
exp_var = pca.explained_variance_ratio_
loadings = pd.DataFrame(pca.components_.T, index=vars_pca, columns=["PC1_loading","PC2_loading"])

print("Varianza explicada:")
print(pd.Series(exp_var, index=["PC1","PC2"]).map(lambda v: f"{v*100:.2f}%"))
print("\nLoadings (contribución de cada variable a PC1 y PC2):")
print(loadings.round(3))

# --------------------------------------------------
# 4) Gráfico de dispersión: scores PC1 vs PC2
#    (coloreamos por año)
# --------------------------------------------------
plt.figure(figsize=(8,6))
if "AÑO" in df_plot.columns:
    for a in sorted(df_plot["AÑO"].dropna().unique()):
        idx = df_plot["AÑO"] == a
        plt.scatter(scores.loc[idx, "PC1"], scores.loc[idx, "PC2"], alpha=0.35, label=f"Año {int(a)}", s=20)
    plt.legend()
else:
    plt.scatter(scores["PC1"], scores["PC2"], alpha=0.35, s=20)

plt.title("PCA (PC1 vs PC2) - Scores")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1) Dataset: edad2 e ITF
# --------------------------------------------------
df_cluster = EPH_limpio[["edad2", "ITF"]].copy()

# Outliers: filtramos ITF en el percentil 1–99
low, high = df_cluster["ITF"].quantile([0.01, 0.99])
df_cluster = df_cluster[(df_cluster["ITF"].between(low, high))]

# Dropeamos nulos
df_cluster = df_cluster.dropna()

# Normalización
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster)

# --------------------------------------------------
# 2) K-means con distintos k
# --------------------------------------------------
for k in [2, 4, 10]:
    kmeans = KMeans(n_clusters=k, n_init=20, random_state=0)
    clusters = kmeans.fit_predict(X_scaled)

    # Guardamos etiquetas en el dataframe
    df_cluster[f"cluster_{k}"] = clusters

    # --------------------------------------------------
    # 3) Gráfico
    # --------------------------------------------------
    plt.figure(figsize=(7,6))
    plt.scatter(df_cluster["edad2"], df_cluster["ITF"],
                c=clusters, cmap="tab10", alpha=0.5)

    # Centroides (en espacio original)
    centers = scaler.inverse_transform(kmeans.cluster_centers_)
    plt.scatter(centers[:,0], centers[:,1], c="red", marker="x", s=200, label="Centroides")

    plt.title(f"K-means con k={k}")
    plt.xlabel("Edad^2")
    plt.ylabel("Ingreso total familiar (ITF)")
    plt.legend()
    plt.show()

```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Variables: edad2 e ITF
df_cluster = EPH_limpio[["edad2", "ITF"]].copy()

# Filtrar outliers de ITF (percentil 1-99)
low, high = df_cluster["ITF"].quantile([0.01, 0.99])
df_cluster = df_cluster[df_cluster["ITF"].between(low, high)].dropna()

# Normalización
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster)

# Calcular inercia para k = 1...10
inertia = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, n_init=20, random_state=0)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# Gráfico del codo
plt.figure(figsize=(7,5))
plt.plot(K_range, inertia, marker='o')
plt.xticks(K_range)
plt.xlabel("Número de clusters (k)")
plt.ylabel("Inercia (disimilitud dentro de los clusters)")
plt.title("Método del codo - KMeans")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Dataset con las 6 variables
vars_clust = ["edad", "edad2", "educ", "ITF", "MIEMBROS_HOGAR", "horastrab"]
df_hclust = EPH_limpio[vars_clust].dropna().astype(float)

# Normalización
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_hclust)

# Clustering jerárquico (método Ward)
Z = linkage(X_scaled, method="ward")

# Dendrograma
plt.figure(figsize=(12, 6))
dendrogram(Z, truncate_mode="level", p=5)  # truncamos para mostrar solo los últimos niveles
plt.title("Dendrograma - Clustering jerárquico (Ward)")
plt.xlabel("Observaciones agrupadas")
plt.ylabel("Distancia (disimilitud)")
plt.show()
```
